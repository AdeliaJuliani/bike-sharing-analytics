## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

### Cleaning Data

#### Fixing ... problem

```python
# Mengubah tipe data kolom 'dteday' menjadi datetime pada kedua DataFrame
day_df["dteday"] = pd.to_datetime(day_df["dteday"])
hour_df["dteday"] = pd.to_datetime(hour_df["dteday"])

# Memverifikasi perubahan tipe data
print("Tipe data dteday pada day_df :", day_df["dteday"].dtype)
print("Tipe data dteday pada hour_df:", hour_df["dteday"].dtype)
```

```python
# Mendefinisikan dictionary mapping untuk setiap kolom kategorikal agar mudah dipahami
season_map = {1: "Spring", 2: "Summer", 3: "Fall", 4: "Winter"}
year_map = {0: 2011, 1: 2012}
workingday_map = {0: "Holiday/Weekend", 1: "Working Day"}
weathersit_map = {
    1: "Clear/Partly Cloudy",
    2: "Mist/Cloudy",
    3: "Light Snow/Rain",
    4: "Heavy Rain/Ice"
}
weekday_map = {
    0: "Sunday", 1: "Monday", 2: "Tuesday", 3: "Wednesday",
    4: "Thursday", 5: "Friday", 6: "Saturday"
}
month_map = {
    1: "Jan", 2: "Feb", 3: "Mar", 4: "Apr", 5: "May", 6: "Jun",
    7: "Jul", 8: "Aug", 9: "Sep", 10: "Oct", 11: "Nov", 12: "Dec"
}

# Menerapkan mapping ke day_df dan hour_df menggunakan loop
for df in [day_df, hour_df]:
    df["season"] = df["season"].map(season_map)
    df["yr"] = df["yr"].map(year_map)
    df["workingday"] = df["workingday"].map(workingday_map)
    df["weathersit"] = df["weathersit"].map(weathersit_map)
    df["weekday"] = df["weekday"].map(weekday_map)
    df["mnth"] = df["mnth"].map(month_map)

# Mengubah nama kolom agar lebih intuitif saat analisis dan pembuatan dashboard
rename_cols = {
    "dteday": "date",
    "yr": "year",
    "mnth": "month",
    "hr": "hour",
    "weathersit": "weather",
    "cnt": "total_rentals"
}
day_df.rename(columns=rename_cols, inplace=True)
hour_df.rename(columns=rename_cols, inplace=True)

# Memeriksa hasil akhir cleaning pada 5 baris pertama hour_df
hour_df.head()
```

### Pertanyaan 1: Bagaimana perbandingan rata-rata jumlah penyewaan sepeda harian antara pengguna terdaftar (registered) dan pengguna kasual (casual) pada hari kerja (working day) dibandingkan hari libur/akhir pekan selama periode 2011 hingga 2012?

```python
# Menyiapkan data rata-rata penyewaan casual vs registered berdasarkan workingday
avg_user_df = day_df.groupby("workingday")[["casual", "registered"]].mean().reset_index()

# Mengubah format tabel menjadi long-format agar mudah divisualisasikan dengan Seaborn
avg_user_melted = avg_user_df.melt(
    id_vars="workingday",
    value_vars=["casual", "registered"],
    var_name="User Type",
    value_name="Average Daily Rentals"
)
avg_user_melted["User Type"] = avg_user_melted["User Type"].str.capitalize()

# Membuat kanvas grafik dengan ukuran proporsional
plt.figure(figsize=(10, 6))

# Membuat grouped bar chart dengan warna kontras yang mudah dibedakan
palette_colors = {"Casual": "#72BCD4", "Registered": "#1F4E78"}
ax = sns.barplot(
    data=avg_user_melted,
    x="workingday",
    y="Average Daily Rentals",
    hue="User Type",
    palette=palette_colors
)

# Menambahkan label angka eksak di atas setiap batang grafik
for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=4, fontsize=11, fontweight="bold")

# Mengatur judul yang menyampaikan insight utama serta label sumbu yang jelas
plt.title(
    "Registered Users Dominate Working Days, While Casual Users Double on Weekends (2011–2012)",
    fontsize=13,
    fontweight="bold",
    pad=15
)
plt.xlabel("Day Category", fontsize=11, fontweight="bold")
plt.ylabel("Average Daily Rentals", fontsize=11, fontweight="bold")
plt.ylim(0, 4600)
plt.legend(title="User Type", fontsize=10, title_fontsize=11)
sns.despine(top=True, right=True)

plt.tight_layout()
plt.show()
```

## Analisis Lanjutan (Opsional)

Clustering Permintaan Sewa (Manual Grouping)

Untuk memberikan wawasan operasional yang lebih mendalam dalam pengelolaan stok sepeda di lapangan, dilakukan analisis pengelompokan (clustering) tanpa algoritma machine learning menggunakan teknik Manual Grouping (Binning). Pengelompokan ini bertujuan mengklasifikasikan setiap jam operasional ke dalam tiga tingkat intensitas permintaan (Low Demand, Medium Demand, dan High Demand) berdasarkan distribusi kuartil penyewaan sepeda serta memetakannya terhadap segmen waktu dalam sehari (Time of Day).

```python
# Membuat salinan DataFrame agar tidak mengubah data utama
clustering_df = hour_df.copy()

# 1. Manual Grouping berdasarkan Segmen Waktu (Time of Day)
def categorize_time_of_day(hour):
    if 0 <= hour < 6:
        return "Late Night (00:00-05:59)"
    elif 6 <= hour < 12:
        return "Morning (06:00-11:59)"
    elif 12 <= hour < 18:
        return "Afternoon (12:00-17:59)"
    else:
        return "Evening (18:00-23:59)"

clustering_df["time_segment"] = clustering_df["hour"].apply(categorize_time_of_day)

# 2. Manual Grouping (Binning) Tingkat Permintaan (Demand Cluster) berdasarkan distribusi kuartil (Q1 & Q3)
# Q1 (~40 sewa/jam) sebagai batas Low Demand, Q3 (~281 sewa/jam) sebagai batas High Demand
q1 = clustering_df["total_rentals"].quantile(0.25)
q3 = clustering_df["total_rentals"].quantile(0.75)

def categorize_demand(rentals):
    if rentals <= q1:
        return "Low Demand"
    elif rentals <= q3:
        return "Medium Demand"
    else:
        return "High Demand"

clustering_df["demand_cluster"] = clustering_df["total_rentals"].apply(categorize_demand)

# Membuat tabel ringkasan distribusi cluster berdasarkan segmen waktu
cluster_summary = pd.crosstab(
    index=clustering_df["time_segment"],
    columns=clustering_df["demand_cluster"],
    margins=True,
    margins_name="Total Hours"
)[["Low Demand", "Medium Demand", "High Demand", "Total Hours"]]

# Mengurutkan indeks segmen waktu secara kronologis
segment_order = [
    "Late Night (00:00-05:59)",
    "Morning (06:00-11:59)",
    "Afternoon (12:00-17:59)",
    "Evening (18:00-23:59)",
    "Total Hours"
]
cluster_summary = cluster_summary.reindex(segment_order)

print(f"Batas Kuartil: Low Demand (<= {q1:.0f} sewa/jam) | High Demand (> {q3:.0f} sewa/jam)\n")
cluster_summary
```

```python
# Menghitung persentase distribusi demand cluster pada setiap segmen waktu
cluster_pct = pd.crosstab(
    index=clustering_df["time_segment"],
    columns=clustering_df["demand_cluster"],
    normalize="index"
)[["Low Demand", "Medium Demand", "High Demand"]] * 100

cluster_pct = cluster_pct.reindex([
    "Late Night (00:00-05:59)",
    "Morning (06:00-11:59)",
    "Afternoon (12:00-17:59)",
    "Evening (18:00-23:59)"
])

# Membuat stacked bar chart untuk memvisualisasikan proporsi cluster di tiap segmen waktu
ax = cluster_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(11, 6),
    color={"Low Demand": "#BDC3C7", "Medium Demand": "#5DADE2", "High Demand": "#1F4E78"},
    edgecolor="white"
)

# Menambahkan label persentase pada tiap segmen batang
for c in ax.containers:
    ax.bar_label(c, fmt="%.1f%%", label_type="center", color="white", fontweight="bold", fontsize=10)

plt.title("Proportion of Rental Demand Clusters by Time of Day (2011–2012)", fontsize=13, fontweight="bold", pad=15)
plt.xlabel("Time Segment", fontsize=11, fontweight="bold")
plt.ylabel("Percentage of Operational Hours (%)", fontsize=11, fontweight="bold")
plt.xticks(rotation=15)
plt.legend(title="Demand Cluster", bbox_to_anchor=(1.02, 1), loc="upper left")
sns.despine(top=True, right=True)

plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Konsentrasi High Demand: Sebanyak 46,1% dari total jam operasional pada segmen Afternoon (12:00–17:59) dan 30,5% pada segmen Evening (18:00–23:59) masuk dalam kategori High Demand (> 281 penyewaan/jam). Ini membuktikan bahwa tekanan beban armada tertinggi terpusat dari siang hingga awal malam.
- Dominasi Low Demand: Segmen Late Night (00:00–05:59) didominasi oleh Low Demand sebesar 78,5% (<= 40 penyewaan/jam), menjadikannya jendela waktu paling aman untuk melakukan penarikan sepeda rusak, pengisian daya sistem stasiun, dan redistribusi stok ke titik komuter tanpa mengganggu ketersediaan layanan.
